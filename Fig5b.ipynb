```python
import numpy as np
import pandas as pd
from shapely.geometry import Point, Polygon

# --- 0. Data preprocessing ---
subset.var_names = [str(g).upper() for g in subset.var_names]
score_set_clean = {k: [g.upper() for g in v] for k, v in score_set.items()}

coords = subset.obsm['spatial']
points = [Point(x, y) for x, y in coords]

# Temporary storage for data (Key: (Type, Orig_ID), Value: dict)
results = {}

# Predefine all target genes
all_target_genes = [g for genes in score_set_clean.values() for g in genes]
unique_genes_list = list(dict.fromkeys(all_target_genes))

# --- Helper function: compute mean values for a single ROI ---
def calculate_raw_roi_data(poly_coords, roi_type, orig_id):
    poly = Polygon(poly_coords)
    mask = np.array([poly.contains(p) for p in points])
    indices = np.where(mask)[0]
    
    if len(indices) == 0:
        return None
    
    sub_X = subset[indices, :].X
    if hasattr(sub_X, 'toarray'): 
        sub_X = sub_X.toarray()
    
    row_data = {'Type': roi_type, 'Orig_ID': orig_id}
    
    # 1. Compute score (mean of the gene set)
    for score_name, genes in score_set_clean.items():
        valid_genes = [g for g in genes if g in subset.var_names]
        if valid_genes:
            gene_indices = [subset.var_names.get_loc(g) for g in valid_genes]
            # Compute raw mean here
            row_data[score_name] = float(sub_X[:, gene_indices].mean())
        else:
            row_data[score_name] = 0.0
            
    # 2. Compute mean expression for individual genes
    for g in unique_genes_list:
        if g in subset.var_names:
            idx = subset.var_names.get_loc(g)
            # Compute raw mean here
            row_data[g] = float(sub_X[:, idx].mean())
        else:
            row_data[g] = 0.0
            
    return row_data

# --- 1. Iterate over Mature (M) ---
for i, poly_coords in enumerate(Mature):
    data = calculate_raw_roi_data(poly_coords, 'M', i+1)
    if data:
        results[('M', i+1)] = data

# --- 2. Iterate over Involution_like (I) ---
for i, poly_coords in enumerate(Involution_like):
    data = calculate_raw_roi_data(poly_coords, 'I', i+1)
    if data:
        results[('I', i+1)] = data

if not results:
    raise ValueError("No cells were extracted!")

# --- 3. Global normalization ---
# Goal: for each gene/score, the sum across all TLS equals FIXED_SUM (e.g. 100)
FIXED_SUM = 100.0 

normalized_results = []

# Get all columns that need normalization (scores + genes)
columns_to_normalize = list(score_set_clean.keys()) + unique_genes_list

for key, row_data in results.items():
    new_row = row_data.copy()
    
    for col in columns_to_normalize:
        val = row_data[col]
        
        # Compute the sum of this column across all TLS
        total_sum = sum(data[col] for data in results.values())
        
        if total_sum > 0:
            # Normalization formula: new value = old value * (fixed sum / current sum)
            new_row[col] = val * (FIXED_SUM / total_sum)
        else:
            # If the gene/score is 0 across all TLS, keep it as 0
            new_row[col] = 0.0
            
    normalized_results.append(new_row)

# --- 4. Build DataFrame and sort ---
df_scores = pd.DataFrame(normalized_results)

# Sort by mature TLS score (normalized) in descending order
df_scores = df_scores.sort_values(by="mature TLS score", ascending=False).reset_index(drop=True)

# Generate Display_ID
df_scores['Display_ID'] = df_scores.apply(lambda x: f"{x['Type']}-{x['Orig_ID']}", axis=1)
df_scores.set_index('Display_ID', inplace=True)

# Prepare heatmap order
group1 = score_set_clean["mature TLS score"]
group2 = score_set_clean["plasma output score"]
group3 = score_set_clean["remodeling/myeloid score"]
heatmap_order = group1 + group2 + group3
seen = set()
unique_heatmap_order = [x for x in heatmap_order if not (x in seen or seen.add(x))]

# Heatmap data also uses the normalized df_scores
df_heatmap = df_scores[unique_heatmap_order]

print(f"Normalization complete. The sum of each gene/score across all TLS is approximately {FIXED_SUM}")
print("Verification (sum of the first 3 genes):")
for g in unique_heatmap_order[:3]:
    print(f"  {g}: {df_scores[g].sum():.2f}")

print("Data preparation complete. First 5 label examples:", df_scores.index[:5].tolist())
```

```python
import matplotlib.gridspec as gridspec
from matplotlib.colors import LinearSegmentedColormap
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# --- Font and rendering settings ---
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['font.size'] = 16
plt.rcParams['axes.linewidth'] = 0.8
plt.rcParams['figure.dpi'] = 150 
sns.set_style("whitegrid", {"grid.color": ".9"})

# --- Data preparation ---
if 'Display_ID' in df_scores.columns:
    df_scores_reset = df_scores.copy()
    df_scores_reset['Display_ID_Short'] = df_scores_reset['Display_ID'].str.replace('-', '')
    df_scores_reset.set_index('Display_ID_Short', inplace=True)
elif not df_scores.index.str.contains('-').any():
    df_scores_reset = df_scores.copy()
else:
    df_scores_reset = df_scores.copy()
    df_scores_reset.index = df_scores_reset.index.str.replace('-', '')

tls_ids = df_scores_reset.index.tolist() 
n_tls = len(tls_ids)
x_pos = np.arange(n_tls)

fig = plt.figure(figsize=(16, 12))
gs = gridspec.GridSpec(4, 1, height_ratios=[0.7, 0.7, 0.7, 2.5], hspace=0.02)

cmap_gray = LinearSegmentedColormap.from_list("gray_scale", ["#f0f0f0", "#000000"])
cmap_purple = LinearSegmentedColormap.from_list("purple_grad", 
                                                ["#fce4ec", "#e1bee7", "#ba68c8", "#8e24aa", "#4a148c"])

def plot_score_bar_aligned(ax, values, ylabel, color_map=cmap_gray):
    bars = ax.bar(x_pos, values, width=1.0, edgecolor='none', linewidth=0, align='center')
    max_val = values.max() if values.max() > 0 else 1.0
    norm_vals = values / max_val
    for i, bar in enumerate(bars):
        intensity = norm_vals[i]
        bar.set_facecolor(color_map(intensity))
        text_color = 'white' if intensity > 0.5 else 'black'
        ax.text(i, values[i]/2, f"{values[i]:.1f}", ha='center', va='center', fontsize=10, fontweight='bold', color=text_color)
    ax.set_ylabel(ylabel, rotation=0, ha='right', fontsize=16, fontweight='bold')
    ax.yaxis.set_label_coords(-0.01, 0.25) 
    ax.set_xticks([])
    ax.set_yticks([])
    for spine in ax.spines.values():
        spine.set_visible(False)
    ax.set_ylim(0, max_val * 1.25)
    ax.set_xlim(-0.5, n_tls - 0.5)

# --- 1, 2, 3. Score bar plots ---
ax1 = fig.add_subplot(gs[0])
plot_score_bar_aligned(ax1, df_scores_reset["mature TLS score"].values, "mature\nTLS score")
ax2 = fig.add_subplot(gs[1])
plot_score_bar_aligned(ax2, df_scores_reset["plasma output score"].values, "plasma\noutput score")
ax3 = fig.add_subplot(gs[2])
plot_score_bar_aligned(ax3, df_scores_reset["remodeling/myeloid score"].values, "remodeling/\nmyeloid score")

# --- Bottom heatmap ---
ax4 = fig.add_subplot(gs[3])

heatmap_data = df_scores_reset[unique_heatmap_order].T 
n_genes = len(unique_heatmap_order)

# Generate grid coordinates
# X: column boundaries (-0.5 to n_tls-0.5+1)
# Y: row boundaries (-0.5 to n_genes-0.5+1)
X, Y = np.meshgrid(np.arange(-0.5, n_tls + 0.5), np.arange(-0.5, n_genes + 0.5))

# Flip the data to match the pcolormesh coordinate system (bottom to top)
# The first row of the original data is at the top, while pcolormesh places it at the bottom, so we flip it
data_to_plot = heatmap_data.values[::-1, :] 

# Draw pcolormesh
im = ax4.pcolormesh(X, Y, data_to_plot, 
                    cmap=cmap_purple, 
                    shading='flat', 
                    edgecolors='white',   # ← Restore white borders
                    linewidth=1.5,        # ← Restore border width (corresponds to previous gridlines)
                    rasterized=True)      # ← Keep vectorization optimization

# Set axis labels
# X-axis remains unchanged
ax4.set_xticks(np.arange(n_tls))
ax4.set_xticklabels(tls_ids, fontsize=14, fontweight='bold', rotation=45)

# Y-axis needs to correspond to the flipped data indices
ax4.set_yticks(np.arange(n_genes)) 
ax4.set_yticklabels(unique_heatmap_order[::-1], fontsize=16) # Labels also need to be flipped to match the flipped data rows

# Limit display range to ensure alignment
ax4.set_xlim(-0.5, n_tls - 0.5)
ax4.set_ylim(-0.5, n_genes - 0.5)

for spine in ax4.spines.values():
    spine.set_visible(False)


# --- Left-side grouping bands ---
g1_len = len(score_set_clean["mature TLS score"])
g2_len = len(score_set_clean["plasma output score"])
g3_len = len(score_set_clean["remodeling/myeloid score"])

# Note: The y-coordinates here are calculated based on the "unflipped" logical order
# However, in the flipped plot:
# The bottom is g3 (Remodeling), the middle is g2 (Plasma), and the top is g1 (Follicular)
# So we need to recalculate the span positions, or draw directly based on the flipped indices

# Flipped index distribution:
# 0 ~ g3_len-1 : Remodeling (originally the last g3_len rows)
# g3_len ~ g3_len+g2_len-1 : Plasma
# g3_len+g2_len ~ end : Follicular

y3_start_flip, y3_end_flip = -0.5, g3_len - 0.5
y2_start_flip, y2_end_flip = g3_len - 0.5, g3_len + g2_len - 0.5
y1_start_flip, y1_end_flip = g3_len + g2_len - 0.5, n_genes - 0.5

# Draw background color blocks (note the changed order)
ax4.axhspan(y3_start_flip, y3_end_flip, color='#8e24aa', alpha=0.3, zorder=0) # Remodeling at the bottom
ax4.axhspan(y2_start_flip, y2_end_flip, color='#ba68c8', alpha=0.3, zorder=0) # Plasma in the middle
ax4.axhspan(y1_start_flip, y1_end_flip, color='#e1bee7', alpha=0.3, zorder=0) # Follicular at the top

# --- Group separator lines (thick black lines) ---
# Draw lines in the flipped coordinate system
ax4.axhline(y=g3_len - 0.5, color='black', linewidth=2, zorder=1) # Between Remodeling and Plasma
ax4.axhline(y=g3_len + g2_len - 0.5, color='black', linewidth=2, zorder=1) # Between Plasma and Follicular

# --- Left-side text labels ---
# Calculate center positions after flipping
cy3_flip = (y3_start_flip + y3_end_flip) / 2
cy2_flip = (y2_start_flip + y2_end_flip) / 2
cy1_flip = (y1_start_flip + y1_end_flip) / 2

text_props = dict(fontsize=16, fontweight='bold', rotation=90, va='center', ha='center')

# Note: The current order from bottom to top is: Remodeling, Plasma, Follicular
# So the text should be placed accordingly
ax4.text(-5, cy3_flip, "Remodeling\nand myeloid\nmarkers", **text_props)
ax4.text(-5, cy2_flip, "Plasma\noutput\nmarkers", **text_props)
ax4.text(-5, cy1_flip,  "Follicular\nmaturity\nmarkers", **text_props)

fig.suptitle("A Continuum of 40 Tertiary Lymphoid Structures from a Post-Treatment pCR Patient:\nFrom Follicular Maturity to Involuting, Remodeling TLSs", fontsize=16, fontweight='bold', y=0.89)

cbar_ax = fig.add_axes([0.92, 0.15, 0.01, 0.7])
fig.colorbar(im, cax=cbar_ax, orientation='vertical')
cbar_ax.set_ylabel('Norm. Log Mean Expr.', rotation=270, labelpad=15, fontsize=16)

plt.subplots_adjust(left=0.1)
plt.show()
```
