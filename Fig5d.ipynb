```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.path import Path

# ================= 1. Data Preparation =================
# Detector for whether a spot is inside multiple polygonal regions
def create_path_from_polygons(polygons):
    vertices = []
    codes = []
    for poly in polygons:
        p_arr = np.array(poly)
        if len(p_arr) < 3: continue
        vertices.append(p_arr)
        c_arr = np.ones(len(p_arr)) * Path.LINETO
        c_arr[0] = Path.MOVETO
        c_arr[-1] = Path.CLOSEPOLY
        codes.append(c_arr)
    return Path(np.vstack(vertices), np.concatenate(codes))
```

```python
import scanpy as sc
adata_HD = sc.read("./h5ad/adata_HD.h5ad")
subset = adata_HD[adata_HD.obs["batch"]=="nS06"]
```

```python
coords = subset.obsm['spatial']
path_mature = create_path_from_polygons(Mature)
path_involution_like = create_path_from_polygons(Involution_like)
is_in_mature = path_mature.contains_points(coords)
is_in_involution_like = path_involution_like.contains_points(coords)
```

```python
n_mature_total = np.sum(is_in_mature)
n_involution_like_total = np.sum(is_in_involution_like)
n_defined_total = n_mature_total + n_involution_like_total
baseline_ratio = n_mature_total / n_defined_total
```

```python
results = []
for gene in ['CR1', 'CR2', 'FCER2', 'TOP2A', 'MKI67']:
    expr = subset[:, gene].X.toarray().flatten() if hasattr(subset[:, gene].X, "toarray") else np.array(subset[:, gene].X).flatten()
    is_pos = expr > 0
    pos_mature = np.sum(is_pos & is_in_mature)
    pos_defined = np.sum(is_pos & (is_in_mature | is_in_involution_like))
    ratio = (pos_mature / pos_defined) 
    results.append({'Gene': gene, 'Ratio': ratio, 'Count': pos_defined})
```
