```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import scanpy as sc

# ================= 1. Data preparation and labeling =================

adata_HD = sc.read("../h5ad/adata_HD.h5ad")
subset = adata_HD[adata_HD.obs["batch"] == "nS06"]
spots = subset.obsm['spatial']
spot_types = subset.obs['type']
color_mapping_dict = {
    'B cell': '#ff7f0e',          
    'Endothelial cell': '#1f77b4',
    'Epithelial cell': '#7f7f7f', 
    'Stromal cell': '#9467bd',  
    'Muscle cell': '#8c564b',     
    'Myeloid cell': '#d62728',    
    'Plasma cell': '#bcbd22',     
    'T cell': '#17becf'           
}

region_labels = ['Unknown'] * len(spots)

def point_in_polygon(point, polygon):
    x, y = point
    inside = False
    p1x, p1y = polygon[0]
    for i in range(1, len(polygon) + 1):
        p2x, p2y = polygon[i % len(polygon)]
        if y > min(p1y, p2y):
            if y <= max(p1y, p2y):
                if x <= max(p1x, p2x):
                    if p1y != p2y:
                        xinters = (y - p1y) * (p2x - p1x) / (p2y - p1y) + p1x
                    if p1x == p2x or x <= xinters:
                        inside = not inside
        p1x, p1y = p2x, p2y
    return inside

# --- Label Mature TLS (M1, M2...) ---
tls_id = 1
for region in Mature:
    if len(region) == 0: continue
    pts = np.array(region)
    closed_pts = np.vstack([pts, pts[0]])
    for i, spot in enumerate(spots):
        if (closed_pts[:, 0].min() <= spot[0] <= closed_pts[:, 0].max() and 
            closed_pts[:, 1].min() <= spot[1] <= closed_pts[:, 1].max()):
            if point_in_polygon(spot, closed_pts[:-1]):
                region_labels[i] = f"M{tls_id}"
    tls_id += 1

# --- Label Involution_like TLS (I1, I2...) ---
for region in Involution_like:
    if len(region) == 0: continue
    pts = np.array(region)
    closed_pts = np.vstack([pts, pts[0]])
    for i, spot in enumerate(spots):
        if (closed_pts[:, 0].min() <= spot[0] <= closed_pts[:, 0].max() and 
            closed_pts[:, 1].min() <= spot[1] <= closed_pts[:, 1].max()):
            if point_in_polygon(spot, closed_pts[:-1]):
                region_labels[i] = f"I{tls_id}"
    tls_id += 1

# ================= 2. Calculate percentage composition =================
df = pd.DataFrame({'Region': region_labels, 'CellType': spot_types})
df = df[df['Region'] != 'Unknown']
pivot_counts = df.pivot_table(index='Region', columns='CellType', aggfunc='size', fill_value=0)
pivot_percent = pivot_counts.div(pivot_counts.sum(axis=1), axis=0) * 100
pivot_percent = pivot_percent.sort_index()

# ================= 3. Plot the refined chart (core modification part) =================

# 1. Create an extra-wide canvas
fig, ax = plt.subplots(figsize=(25, 8))

# 2. Prepare data
x_positions = np.arange(len(pivot_percent))  # x-axis position of each bar
ordered_labels = ["B cell","Endothelial cell","Epithelial cell","Fibroblast",
                  "Muscle cell","Myeloid cell","Plasma cell","T cell"]
# Extract corresponding colors
bar_colors = [color_mapping_dict[col] for col in pivot_percent.columns]

# 3. Manually draw the stacked bar chart using ax.bar()
bottom_heights = np.zeros(len(pivot_percent))  # starting height for each stacked layer
for i, cell_type in enumerate(pivot_percent.columns):
    ax.bar(x_positions, 
           pivot_percent[cell_type], 
           bottom=bottom_heights,  # bottom of the stack
           color=bar_colors[i],    # use custom colors
           edgecolor='none',       # remove borders
           label=cell_type)        # legend label
    # Update the starting height for the next layer
    bottom_heights += pivot_percent[cell_type]

# 4. Styling settings
ax.set_ylabel("Composition (%)", fontsize=18)
ax.set_title("Cell Composition of TLS Regions", fontsize=20, pad=-2)
ax.set_xticks(x_positions)
ax.set_xticklabels(pivot_percent.index, rotation=0, fontsize=16)  # X-axis labels
ax.set_yticklabels(ax.get_yticklabels(), fontsize=16)
ax.grid(axis='y', linestyle='--', alpha=0.4)

# 5. Legend settings (no border)
ax.legend(title="Cell Type", 
          bbox_to_anchor=(0.97, 0.75), 
          loc='upper left',
          frameon=False,    # Key: remove legend border
          fontsize=16,
          title_fontsize=18)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```
