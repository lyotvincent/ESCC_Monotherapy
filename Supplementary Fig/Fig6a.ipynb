```python
import scanpy as sc
import matplotlib.pyplot as plt
import numpy as np

adata_fibroblast = sc.read("../h5ad/adata_fibroblast.h5ad")

# Set global font and style for a more professional look
plt.rcParams.update({
    'font.size': 12,
    'axes.linewidth': 1.2,
    'xtick.major.width': 1.2,
    'ytick.major.width': 1.2,
    'xtick.direction': 'in',
    'ytick.direction': 'in'
})

fig, ax = plt.subplots(figsize=(8, 4))

sc.pl.umap(
    adata_fibroblast,
    color=["celltype"],
    size=7,
    alpha=1.0,
    show=False,
    ax=ax,
    title=""
)

# Remove top and right borders
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Get legend handles and labels
handles, labels = ax.get_legend_handles_labels()

legend = ax.legend(
    handles,
    labels,
    loc='center left',
    bbox_to_anchor=(1.02, 0.5),
    frameon=False,
    fontsize=10,
    title="Cell Type",
    title_fontsize=12,
    handletextpad=0.4,
    labelspacing=0.6
)

plt.tight_layout()
plt.show()
```
