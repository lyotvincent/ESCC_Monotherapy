```python
from matplotlib.path import Path
import numpy as np

regions = Mature + Involution_like

# Batch Checking
mask = np.column_stack([
    Path(reg).contains_points(spots) 
    for reg in regions
])
```

```python
subset.obs["Mature_TLS"] = [(np.any(row[:24]) & (~np.any(row[24:]))) for row in mask]
subset.obs["Involution_like_TLS"] = [(np.any(row[24:]) & (~np.any(row[:24]))) for row in mask]
subset.obs["in_TLS"] = [np.any(row) for row in mask]
subset.obs["group"] = "other"
subset.obs.loc[subset.obs["Mature_TLS"]==True,"group"] = "Mature"
subset.obs.loc[subset.obs["Involution_like_TLS"]==True,"group"] = "Involution_like"
```

```python
subset.obs["group"] = subset.obs["group"].astype("category")
tp = subset[subset.obs["in_TLS"]].copy()

#Normalized Log Expression 
sc.pp.normalize_total(tp, target_sum=1e6)
sc.pp.log1p(tp)   
# DEG
sc.tl.rank_genes_groups(
    tp,
    groupby="group",
    groups=["Mature"],
    reference="Involution_like",
    method="wilcoxon"
)

# Extraction Results
deg_df = sc.get.rank_genes_groups_df(tp, group="Mature")
deg_df["log2fc"] = deg_df["logfoldchanges"] / np.log(2)

# Save
sig = deg_df[(deg_df["pvals_adj"] < 0.05) & (np.abs(deg_df["log2fc"]) > 1.0)]

sig.sort_values("pvals_adj").to_csv("./table/DEG_Mature_vs_Involution_like_FDR.csv", index=False)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import ListedColormap

# ==========================================
# [New] 0. Manual offset configuration for gene labels
# ==========================================
MANUAL_OFFSETS = {
     'RGS1': (0, 4),   
     'IGHG1': (-0.13, 5),   
     'IGHG3': (0.25, 5),
     'LYZ':(0.3,0),
     'C1QA':(0,-20),
     'JCHAIN':(-0.5,-7),
     "MZB1":(-0.5,-2)
}
# ==========================================

# ==========================================
# 1. Data loading and preprocessing
# ==========================================
try:
    df = pd.read_csv('./table/DEG_Mature_vs_Involution_like_FDR.csv')
except FileNotFoundError:
    print("Error: File not found. Generating simulated data for demonstration...")
    df = pd.DataFrame({
        'names': [f'GENE{i}' for i in range(500)],
        'log2fc': np.random.normal(0, 2, 500),
        'pvals': np.random.uniform(0, 1, 500)
    })
    key_genes = ['MZB1', 'LYZ', 'CD19', 'LTF']
    for i, g in enumerate(key_genes):
        df.loc[i, 'names'] = g
        df.loc[i, 'log2fc'] = (i - 1.5) * 2 
        df.loc[i, 'pvals'] = 1e-10

df['pvals'] = df['pvals'].replace(0, 1e-300)
df['neg_log10_p'] = -np.log10(df['pvals'])

fc_thresh = 1.0  
p_thresh = 0.05  
y_thresh = -np.log10(p_thresh)

# ==========================================
# 2. Define gene groups and colors
# ==========================================
groups = {
    'Plasma cell / antibody-secreting markers': ['MZB1', 'JCHAIN', 'XBP1', 'IGHG1', 'IGHG3'],
    'Macrophage-enriched markers':  ['LYZ', 'C1QA', 'RGS1'],
    'B-cell markers': ['CD19', 'MS4A1', 'CD22'],
    'FDC-associated markers':  ['CLU', 'FDCSP'],
    "Inflammatory secretory marker":["LTF"]
}

group_colors = {
    'Plasma cell / antibody-secreting markers': '#0072B2', 
    'Macrophage-enriched markers':  '#D55E00', 
    'B-cell markers': '#CC79A7', 
    'FDC-associated markers':  '#009E73',  
    "Inflammatory secretory marker": "#F0E442" 
}

df['category'] = 'Background'

for group_name, gene_list in groups.items():
    mask = df['names'].isin(gene_list)
    df.loc[mask, 'category'] = group_name

# ==========================================
# 3. Plotting
# ==========================================
fig, ax = plt.subplots(figsize=(12, 7))

# --- B. Plot highlighted genes ---
for group_name, color in group_colors.items():
    subset = df[df['category'] == group_name]
    if len(subset) > 0:
        ax.scatter(subset['log2fc'], subset['neg_log10_p'], 
                   c=color, s=200, alpha=0.9, label=group_name, 
                   edgecolors='white', linewidth=0.8, zorder=2)

# --- A. Plot background genes ---
bg_data = df[df['category'] == 'Background']
ax.scatter(bg_data['log2fc'], bg_data['neg_log10_p'], 
           c='#d3d3d3', s=150, alpha=0.6, label='Other Genes', zorder=1)

# --- C. Add reference lines ---
ax.axhline(y=y_thresh, color='gray', linestyle='--', linewidth=1, alpha=0.5)
ax.axvline(x=-fc_thresh, color='gray', linestyle='--', linewidth=1, alpha=0.5)
ax.axvline(x=fc_thresh, color='gray', linestyle='--', linewidth=1, alpha=0.5)

# --- D. Add gene labels ---
label_data = df[df['category'] != 'Background']
for _, row in label_data.iterrows():
    gene_name = row['names']
    
    ha = 'right' if row['log2fc'] < 0 else 'left'
    va = 'bottom'
    offset_x = 0
    offset_y = 0
    
    if gene_name in MANUAL_OFFSETS:
        offset_x, offset_y = MANUAL_OFFSETS[gene_name]
        ha = 'center' 
    
    ax.text(row['log2fc'] + offset_x, row['neg_log10_p'] + offset_y, gene_name, 
            fontsize=16, fontweight='bold', ha=ha, va=va,
            color='black')

# --- E. Add region annotations and arrows ---
text_y_pos = y_thresh * 0.3 + 10  
arrow_y_pos = y_thresh * 0.1 

ax.text(-fc_thresh + 1.7, text_y_pos+5, 'Enriched in\nInvoluted TLS\n', 
        ha='right', va='center', fontsize=16, fontweight='bold', color='darkred')
ax.annotate('', xy=(-fc_thresh + 1.8, arrow_y_pos+5), xytext=(-fc_thresh, arrow_y_pos+5),
            arrowprops=dict(arrowstyle='<-', color='darkred', lw=2.5))

ax.text(fc_thresh - 1.7, text_y_pos+40, 'Enriched in\nMature TLS\n', 
        ha='left', va='center', fontsize=16, fontweight='bold', color='darkgreen')
ax.annotate('', xy=(fc_thresh - 1.8, arrow_y_pos+40), xytext=(fc_thresh, arrow_y_pos+40),
            arrowprops=dict(arrowstyle='<-', color='darkgreen', lw=2.5))


# ==========================================
# 4. Styling and output (fixed: legend overlaying the plot)
# ==========================================
ax.set_xlabel('Log₂ Fold Change', fontsize=16, fontweight='bold')
ax.set_ylabel('- Log₁₀ P-value', fontsize=16, fontweight='bold',labelpad=-5)
ax.set_title('TLS DEG: Mature vs Involuted', fontsize=16, fontweight='bold', pad=20)
ax.set_xlim(-3.8, 6.8)

# Y-axis tick processing
current_yticks = ax.get_yticks()
normal_ticks = [t for t in current_yticks if 0 <= t < 300]
new_tick_positions = normal_ticks + [300]
new_tick_labels = [str(int(t)) for t in normal_ticks] + ['≥300']
ax.set_yticks(new_tick_positions)
ax.set_yticklabels(new_tick_labels, fontsize=16)
ax.set_xticklabels([str(int(n)) for n in  ax.get_xticks()],fontsize=16)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# ================= Core modification: legend overlaying the plot =================
legend_obj = ax.legend(
    loc='center right',      # 1. Position: inside the plot, upper-right area
    bbox_to_anchor = (0.97,0.51),
    frameon=True,           # 2. Show border to cover background points for clarity
    facecolor='white',      # 3. White background
    edgecolor='black',      # 4. Black border
    framealpha=0.9,         # 5. Slight transparency (0-1) for a softer look
    fontsize=14,            # 6. Font size
    handletextpad=0.0       # 7. Spacing between marker and text
)

# Make legend text bold
for text in legend_obj.get_texts():
    text.set_fontweight('bold')

# Auto-adjust layout
plt.tight_layout()
plt.show()
```
